### Lists 

A list is a python data structure used to store an ordered collection of elements that can belong to different data types. Lists are dynamic, meaning their size can change during execution. We can use [] to denote a list, and [n] to access it's nth element. 

**(Note: List indexing starts from 0)**

````python
# Create a list
xs = [1, "one", True]

# Print the whole list
print(xs)

# Print one element
print(xs[1])
````

We can also store a list in a lits, giving us access to multi-dimensional storage options. We can access a single element with repeated [].

````python
# Create and print a 2D array
xs_2D = [[1,2],[3,4]]
print(xs_2D)

#print the 1st (0) element of the 2nd (1) Dimension
print(xs_2D[1][0])
````

#### List opperations
Python offers native support for some simple list operations, have a look at the following operators and try to make sense of what they are doing, try to be as precise as possible.

````python
#initialise some test variables
list1 = [1,2]
list2 = [3,4]
elem = 5

#Concatenate two litst
concatenate = list1 + list2 
print(f"concatenate: {concatenate}")

#Append element to the end of a list
list1.append(elem)
print(f"append: {list1}")

#Return the location of the 1st instance of an element
search_for_index = list1.index(elem)
print(f"elem at index: {search_for_index}")

#Insert an element into position 1
list1.insert(1, elem)
print(f"post insert list: {list1}")

#Remove (and return) the last element
pop_elem =  list1.pop()
print(f"post pop list: {list1}")
print(f"popped: {pop_elem}")

#Remove the first element with the specified value
list1.remove(5)
print(f"post remove list: {list1}")

#Return a copy of the list
list_copy = list1.copy()
print(f"copy: {list_copy}")

#reverse the list
list1.reverse()
print(f"reverse: {list1}")

#sort the list
list3 = [5,3,6,1,8]
list3.sort()
print(f"sorted: {list3}")


````

### Tuples

Tuples are similar to lists in that both store a collection objects in sequence however, unlike lists, tuples are **immutable** (We cannot update items to a tuple once it is created).

This restriction allows tuples to be much more efficient than lists, needed less memory and offering faster access times. As you develop bigger and more complex programs it will become important to consider where a tuple may be more appropriate than a list, and vise versa.

````python
# A tuple is declared using () instead of []
tup = (1, 2, 3, 4, 5)

# Tuples are indexed like lists
print(tup[0])
print(tup[4])

# Updating an element of a tuple will lead to an error
tup[1] = 100
print(tup)
````

### Dictionaries

You can think of a python dictionary as a collection of **Key - Value** pairs. They function similarly to an array but values are accessed using the corresponding key instead of the usual index.

````python
cs940dict = {
  "module": "Advanced Algorithms and Programming",
  "code": "cs940",
  "year": 2026,
  "organiser": "Till",
  "tutors" : ["Sid", "Jack"]
}
print(cs940dict["year"])
````

To add new entries to the dictionary, we just index a new key and assign a value:

````python
cs940dict["AvgGrade"] = 65

print(cs940dict)
````

Pandas is an open-source Python library used for data manipulation, analysis and cleaning. It provides fast and flexible tools to work with tabular (csv) data like spreadsheets or SQL tables. It offers good integration with many popular data science tools and libraries, such as numpy.

We can create series using pandas which funciton much like a numpy array, and in fact can be created from numpy arrays, as well as dictionaries, lists and other collections. Pandas series are more complex than numpy arrays but enable more powerful data wrangling and sorting techniques. In general you should use the lightest weight data structure that will meet your requirements. 

An overview of pandas functionality can be found here, take a look but don't worry too much about the details right now.

More useful than series are Pandas Dataframes. These are 2D structures with labeled axis, which can be created by reading directly from a csv (comma separated values) file or SQL database. 

````python
## Code to load a test csv ##

import pandas as pd

#load and print the dataframe
df = pd.read_csv("LabData/lab2_example.csv")
print(df)

print("INFO:")
# Print some informantion about the dataframe including column names, data types, non-null counts, and memory usage.
print(df.info())
````

There are lots of useful operations we can perform on a dataframe, for example:

````python
## Some basic pandas operations ##

#count null values
df.isnull().sum()

#replace null values with a default entry
df = df.fillna(0)

# filtering data
ages = df[df['age'] > 25]
print(ages)

# we can add new columns the same way we woult to a dictionary
df['total'] = df['a'] + df['b']
print(df.head())

# Group data similarly to a SQL database
res = df.groupby('category')['sales'].sum()
print(res)
````

### Generators

Sometimes you may want to create a method which can pause, return a value and resume execution later in the code. We call such functions generators, and they can be very useful in creating your own data structures.

Python supports generators with the 'yield' keyword:

````python
def gen():
  yield 1
  yield 2
  yield 3

print(f"1st call: {gen()}")
print(f"2nd call: {gen()}")
print(f"3rd call: {gen()}")
````
