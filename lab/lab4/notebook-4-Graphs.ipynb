# Graph Representations

This lab will focus the different ways we can represent graphs in python. 

You can think of a graph as a set of "nodes", and a set of "edges" (or connections) between those nodes. All we need to do to implement a graph in python is to come up with a way to record these nodes and edges.


### Adjaceny  Lists

The simplest way to represent a list in python is with an adjaceny list. If two nodes have an edge connecting them we say they are adjacent.

A simple adjaceny list is just a list of all adjacent nodes in the graph. An adjacency list might look somthing like this:

```
adj = [[0, 1], [1, 2], [2, 0], [2,3]]
```

Which describes a graph with nodes 0, 1 & 2. Where:  
0 --> 1  
1 --> 2  
2 --> 0, 3


While this works, it is inefficient and slow to search over. A better solution might be to use a dictionary where the keys are the nodes, and the values are all node adjacent to the key. For example:

```
adj = {0:[1], 1:[2], 2:[0,3]}
```

## Object Oriented Graphs

Another way we can represent graphs is through Object Oriented Programming techniques (OOP). By creating a Node/ Vertices class, which links other nodes, we can define our graph. This will also allow us to store values and write functions within the nodes of our graph, which can be very useful for some applications.

### Task 2.1: OOP Graph Nodes
Complete the following class definitions so that each node can store an integer, as well as a link to any number of neighbours. To test your understanding, you should also include a function which will return the sum of the values stored at the current node and all of its adjacent nodes.
````python
class Node():
    def __init__(self, value):
        self.node = value
        self.link_list = []

    def add_adjacent(self, new_node, weight):
        self.link_list.append((new_node, weight))

    def sum_adjacent(self):
        return sum(weight for neighbor, weight in self.link_list)

    def __repr__(self):
        return str(self.node)
````

## NetworkX
NetworkX is a powerful python library designed to help you work with graphs. We can initialise a graph with the following code (you may need to install networkX with pip first!) 

````python
import networkx as nx
G = nx.Graph()
````

#### NetworkX then allows us to create nodes with function calls on G:

```
#create node '1'
G.add_node(1)

#create nodes 2 and 3
G.add_nodes_from([2, 3])

#create nodes 4 and 5 and give give them colours
G.add_nodes_from([(4, {"color": "red"}), (5, {"color": "green"})])
```

#### We can also add edges in the same way:

```
#Adding a single edge
G.add_edge(1, 2)
e = (2, 3)
G.add_edge(*e)  # unpack edge tuple*

#adding multiple edges
G.add_edges_from([(1, 2), (1, 3)])
```

#### (We can view nodes and edges by accessing G.nodes and G.edges like a dictionary)




#### we can also declare directed graphs with the following:

```
#directed graph
DG = nx.DiGraph()
DG.add_edge(2, 1)   # adds the nodes in order 2, 1
```

#### And finally we can give nodes and edges attributes (like weight) by adding them as arguments to the function calls, this essentially creates an attribute dictionary at the node/ edge

```
#declaring node 1 to have a time attribute
G.add_node(1, time='5pm')

#adding an attribute to an existing node works just like adding a key to a dictionary
G.nodes[1]['room'] = 714


#adding a single edge with a weight
G.add_edge(1, 2, weight=4.7)

#declaring multiple weighted edges
G.add_weighted_edges_from([(1, 2, 0.5), (3, 1, 0.75)])
```

````python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()

G.add_node(1)
G.add_nodes_from([2, 3])
G.add_nodes_from([
    (4, {"color": "red"}),
    (5, {"color": "blue"}),
    (6, {"color": "yellow"}),
    (7, {"color": "black"})
])
G.add_node(10, time="11am")

G.add_edge(1, 2)
G.add_edge(3, 4)

G.nodes[1]["attribute"] = "special"

G.add_edge(2, 3, weight=7.4)
G.add_weighted_edges_from([
    (5, 6, 1.2),
    (6, 7, 2.4)
])

# 生成节点布局
pos = nx.spring_layout(G, seed=42)

# 获取节点颜色，没有 color 属性时使用 lightblue
node_colors = [
    G.nodes[node].get("color", "lightblue")
    for node in G.nodes
]

# 显示节点编号和属性
node_labels = {}

for node, data in G.nodes(data=True):
    details = [str(node)]

    if "attribute" in data:
        details.append(data["attribute"])

    if "time" in data:
        details.append(data["time"])

    node_labels[node] = "\n".join(details)

# 绘制节点和边
nx.draw_networkx_nodes(
    G,
    pos,
    node_color=node_colors,
    node_size=1200
)

nx.draw_networkx_edges(
    G,
    pos,
    width=2
)

# 绘制节点标签
nx.draw_networkx_labels(
    G,
    pos,
    labels=node_labels,
    font_size=10,
    font_color="white"
)

# 获取并绘制边权重
edge_labels = nx.get_edge_attributes(G, "weight")

nx.draw_networkx_edge_labels(
    G,
    pos,
    edge_labels=edge_labels,
    font_size=10,
    label_pos=0.5
)

plt.title("Graph with Node Attributes and Edge Weights")
plt.axis("off")
plt.show()
````
