# Lab 1 Extension: Memory Profiling

Jupyter provides us with some very useful "magic" comamnds which allow us to evaluate the time and memory usage of our code. These commands are:

```
    %%time: Time how long the cell takes to run
    %%timeit: Run the cell several times and report the average time taken
    %%memeit: Run the code several times and report the average memory usage
```

Have a look at the following cells and how %%timeit and %%memit can be used to profile simple loops:

````python
%%timeit

sum = 0
for i in range(100):
    sum += i
````
