### Task 5: A Guessing Game

Now we'll write a simple computer game where we have to guess a number between 1 and
10 that the computer has randomly generated. A sample play through of the game is shown below.
```
I’m thinking of a number, can you guess what it is? 6
Too Low! Guess again. 10
Too High! Guess again. 8
Well done. The number is 8
```

You can use the random number generator in Task 2 to create the target number.

You can take user input with:

```

```

In VSCode, this will open a text box at the top of your page.

For bonus points add in three levels of difficulty which the user can select from:

Level 1 (Easy): A number between 1-5

Level 2 (Medium): A number between 1-10

Level 3 (Hard): a number between 1-100




````python
import random

def response(guessNumber, resultNumber):
    if guessNumber > resultNumber:
        print(f"Too High! Guess again. {guessNumber} ")
        return True;
    elif guessNumber < resultNumber:
        print(f"Too Low! Guess again. {guessNumber} ")
        return True;
    else:
        print(f"Well done. The number is {guessNumber}")
        return False;

level = int(input("choose level: "))
print(f"You entered: {level}")
generateNumber = 0
if level == 1:
    generateNumber = random.randint(0,5)
elif level == 2:
    generateNumber = random.randint(0,10)
else:
    generateNumber = random.randint(0,100)

print(f"number is : {generateNumber}")

inputNumber = int(input("guess a number: "))
while (response(inputNumber, generateNumber)):
    inputNumber = int(input("Guess again: "))



````

### Debugging

There are three main types of errors you are likely to encounter while programming:

1. Syntax errors: refers to the structure and the rules of the programming language. Python will not run the program and terminate immediately with an error message.
2. Runtime error: If there are no syntax errors, the program can start running. If something goes wrong during running (e.g. division by 0), then the program will terminate with a runtime error.
3. Logic error: a program will run fully and terminate, however it will not give the expected result. This is a logic error as there is something wrong with the logic of the program that cannot be picked up by the program.
