## JSON Files

A JSON (JavaScript Object Notation) file is a human readable string format for storing object parameters. You can think of them as a description of a dictionary which can be used to instantiate an object. Although they were made for JavaScript, python provides comprehensive support for JSON files. This will be of particular use for coursework 1, where you will be provided with a set of JSON files to populate your objects.

````python
import json

# some JSON text:
x =  '{ "module":"cs940", "students":90, "organiser":"Till"}'

# load JSON:
y = json.loads(x)

# this gives a Python dictionary:
print(y["students"])
````

We can write an existing dictionary to JSON file with ```json.dump```. This will give you a string containing the JSON encoding, which can then be saved for later use. 

````python
x = {"module":"cs940", "students":90, "organiser":"Till"}

# convert to JSON:
y = json.dumps(x)

# a JSON string:
print(y)
````

To improve readability we can specify how many indentations we want the JSON file to have. Full documentation of Dump and its formatting options can be found [here]

[here]: https://docs.python.org/3/library/json.html

````python
x = {
  "module": "cs940",
  "students": 90,
  "exam": False,
  "coursework": [
    {"name": "cwk1", "weight": 50},
    {"name": "cwk2", "weight": 50}
  ]
}

#default dump
print(json.dumps(x))

#specifying the indent 每一层缩进 2 个空格，让 JSON 更容易阅读。
print(json.dumps(x, indent=2))
````

### Loading an Object from JSON

Since a JSON file is just a dictionary, we can use them to initialise an object by calling the object constructor with the dictionary as the argument. If we use the ** (dict unpack) operator then we convert the dictionary keys directly to function parameters and pass in the values,
take a look at the following code where we create one of our card objects using JSON:

````python
#create our JSON
card =  '{ "suit":"Hearts", "value":7}'
card_json = json.loads(card)

print(card_json)
seven_hearts = lo.Card(**card_json)
print(seven_hearts)

#Compare this to standard initialisation
eight_hearts = lo.Card(suit="Hearts", value=8)
print(eight_hearts)
````

### Task 5.1

Update your Card and Hand objects so that they can be stored as JSON files using the following code. To do this you will need to define the \__dict\__ magic methods for both classes such that they return a dictionary describing the current state of the objects

````python
#create a hand of cards
test_cards = [seven_hearts, eight_hearts]
hand = lo.Hand(cards= test_cards)

# convert to JSON and print results (only for visualisation)
saved_hand = json.dumps(hand.__dict__(), indent=2)
print(saved_hand)

# Save the JSON file for later use
with open('hand.json', 'w', encoding='utf-8') as f:
    json.dump(hand.__dict__(), f, ensure_ascii=False, indent=4)
````
